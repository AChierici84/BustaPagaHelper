```python
class ImageFolderDataset(Dataset):
  def __init__(self, root_dir:str,  split:str="train", transform: callable = None, labels_map: dict= LABEL_TO_TARGET):
    """
    inizializza il dataset
    args:
       root_dir : directory base del dataset
       split: train, val, test
       transform: trasformazioni da applicare all'immagine
       labels_map: dizionario che mappa le classi in numeri

    """
    self.root_dir = root_dir
    self.split = split
    self.split_dir = os.path.join(root_dir, split)
    self.images, self.targets= [],[]
    for label in os.listdir(self.split_dir):
      label_dir = os.path.join(self.split_dir, label)
      if not os.path.isdir(label_dir):
        continue
      for image_name in os.listdir(label_dir):
        # Skip files starting with '._'
        if image_name.startswith('._'):
            continue
        image_path = os.path.join(label_dir, image_name)
        self.images.append(image_path)
        self.targets.append(labels_map[label])
    self.transform = transform
    self.labels_map = labels_map

  def __getitem__(self, index: int)-> tuple: #Torch.tensor,int
    """
    restituisce un'immagine e il suo target
    """
    image = Image.open(self.images[index]).convert("RGB")
    target = self.targets[index]
    if self.transform is not None:
      image = self.transform(image)
    else:
      raise Exception("transform not implemented")
    return image, target

  def get_labels_map(self):
    """
    restituisce il dizionario delle classi
    output:
      labels_map: dizionario che mappa le classi in numeri
    """
    return self.labels_map

  def visualize_transform(self, count: int = 1, mean:tuple =(0.485, 0.456, 0.406), std:tuple =(0.229, 0.224, 0.225), seed: int = RANDOM_SEED):
    """
    Confronta immagini originali e trasformate
    Args:
      count: numero di immagini da visualizzare
      mean: media delle immagini
      std: deviazione standard delle immagini
      seed: seme per il generatore di numeri casuali
    """
    if seed is not None:
        random.seed(seed)

    count = min(count, len(self.images))
    indices = random.sample(range(len(self.images)), count)

    mean = torch.tensor(mean).view(3, 1, 1)
    std = torch.tensor(std).view(3, 1, 1)

    fig, axs = plt.subplots(count, 2, figsize=(10, 5 * count))
    if count == 1:
        axs = [axs]

    for row, idx in enumerate(indices):
        image_transformed, target = self[idx]
        image_original = Image.open(self.images[idx]).convert("RGB")

        # se l'immagine è un tensore la denormalizzo.
        if isinstance(image_transformed, torch.Tensor):
            image_denorm = image_transformed * std + mean
            image_denorm = image_denorm.permute(1, 2, 0).clamp(0, 1)
        else:
            image_denorm = image_transformed

        axs[row][0].imshow(image_original)
        axs[row][0].set_title(f"Original (idx={idx})")
        axs[row][0].axis("off")

        axs[row][1].imshow(image_denorm)
        axs[row][1].set_title(f"Transformed (target={target})")
        axs[row][1].axis("off")

    plt.tight_layout()
    plt.show()

  def class_distribution(self, verbose:bool=True):
    """
    verifica la distruzione delle classi sul dataset
    args:
      verbose: se True stampa le informazioni
    output:
      dist: dizionario che mappa le classi in numero di elementi e percentuale di elementi
    """
    counter = Counter(self.targets)
    total = len(self.targets)

    dist = {}
    for label_id, count in counter.items():
        label_name = [k for k, v in self.labels_map.items() if v == label_id][0]
        dist[label_name] = {
            "count": count,
            "percentage": 100 * count / total
        }

    if verbose:
        print(f"Class distribution ({self.split})")
        for label, info in dist.items():
            print(f"- {label}: {info['count']} ({info['percentage']:.2f}%)")

    return dist

  def plot_class_distribution(self):
    """
    plotta la distribuzione delle classi sul dataset
    """
    counter = Counter(self.targets)
    labels = list(counter.keys())
    counts = list(counter.values())

    label_names = [
        [k for k, v in self.labels_map.items() if v == l][0]
        for l in labels
    ]

    plt.figure(figsize=(6,4))
    plt.bar(label_names, counts)
    plt.title(f"Class distribution - {self.split}")
    plt.ylabel("Number of samples")
    plt.show()

  def __len__(self)-> int:
    return len(self.images)
```

```python
  def compare_splits_distribution(datasets):
    """
    Compara la distribuzione delle classi tra i dataset di train, validation e test.
    Args:
        datasets: dizionario contente i tre dataset di train, validation e test.
    """

    rows = []
    for split, ds in datasets.items():
        dist = ds.class_distribution(verbose=False)
        for label, info in dist.items():
            rows.append({
                "split": split,
                "label": label,
                "count": info["count"],
                "percentage": info["percentage"]
            })

    df = pd.DataFrame(rows)
    return df
```

```python
class EarlyStopping:
  """
  Early stopping per il training
  """
  def __init__(self,patience:int=3,min_delta:float=0):
    """
    Inizializzo i parametri
    args:
         patiance: quante epoche senza miglioramenti devo attendere
         min_delta: quanto devo cambiare per considerare un miglioramento
    """
    self.patience=patience
    self.min_delta=min_delta
    self.counter=0
    self.best_loss=float("inf")
    self.early_stop=False

  def __call__(self,val_loss:float):
    """
    Verifico se devo fermarmi
    args:
        val_loss: loss di validazione
    """
    if val_loss< self.best_loss - self.min_delta:
      self.best_loss = val_loss
      self.counter=0
    else:
      self.counter += 1
      if self.counter >= self.patience:
        self.early_stop=True

class ModelCheckpoint:
    """
    Callback per salvare il miglior modello
    """
    def __init__(self, output_dir: str, filename: str = "best_model.pth"):
        self.output_dir = output_dir
        self.filepath = os.path.join(output_dir, filename)
        self.best_model_path =""
        self.best_loss = float("inf")
        self.best_epoch = -1
        os.makedirs(output_dir, exist_ok=True)

    def __call__(self, val_loss: float, model: nn.Module, epoch: int):
        if val_loss < self.best_loss:
            self.best_loss = val_loss
            self.best_epoch = epoch
            self.best_model_path = self.filepath
            torch.save(model.state_dict(), self.filepath)
            print(f"Model saved at epoch {self.best_epoch} with val loss {self.best_loss}")
        else:
            print(f"Model not saved at epoch {epoch} (validation loss > {self.best_loss} )")
```

```python
class Trainer:
    """
    Classe per il training
    """
    def __init__(self,
                 name:str,
                 model:nn.Module,
                 optimizer: torch.optim.Optimizer,
                 criterion: nn.Module,
                 epochs:int=100,
                 device: torch.device = torch.device("cpu"),
                 log_every_n_iterations:int=10,
                 early_stopping_patience:int=8,
                 early_stopping_delta:float=1e-4,
                 scheduler_patiance:int=2,
                 scheduler_factor:float=0.3,
                 scheduler_min_lr:float=1e-6
                 ):
        """
        Inizializza i parametri
        args:
            name: nome del modello
            model: modello da allenare
            optimizer: ottimizzatore da utilizzare
            criterion: funzione di costo da utilizzare
            epochs: numero di epoche
            device: dispositivo su cui allenare il modello
            log_every_n_iterations: ogni quanti step mostrare le metriche attuali
            early_stopping_patience: quante epoche senza miglioramenti devo attendere
            early_stopping_delta: quanto devo cambiare per considerare un miglioramento
            scheduler_patiance: quante epoche senza miglioramenti devo attendere per diminuire il learning rate
            scheduler_factor: fattore per diminuire il learning rate
            scheduler_min_lr: minimo learning rate
        """
        self.name = name
        self.epochs = epochs
        self.model = model
        self.optimizer = optimizer
        self.criterion = criterion
        self.log_every_n_iterations = log_every_n_iterations
        self.device = device
        self.scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(optimizer, mode="min", patience=scheduler_patiance, factor=scheduler_factor,min_lr=scheduler_min_lr)
        self.train_losses = []
        self.val_losses = []
        self.train_accuracies = []
        self.val_accuracies = []
        self.train_F1=[]
        self.val_F1=[]
        self.lr_values=[]
        self.early_stop = EarlyStopping(patience=early_stopping_patience, min_delta=early_stopping_delta)
        self.model_checkpoint = ModelCheckpoint(output_dir="./models/"+self.name)
        self.model.to(self.device)


    def train(self, train_loader:DataLoader, val_loader:DataLoader):
        """
        Allenamento del modello
        args:
            train_loader(DataLoader): dataloader di training
            val_loader (DataLoader): dataloader di validation
        """
        for epoch in range(self.epochs):
             print(f"Epoch {epoch+1}/{self.epochs}")
             print("Train Report")
             train_loss,train_accuracy,train_f1=self.train_epoch(train_loader)
             self.train_losses.append(train_loss)
             self.train_accuracies.append(train_accuracy)
             self.train_F1.append(train_f1)
             print("Validation Report")
             val_loss,val_accuracy, val_f1= self.val_epoch(val_loader)
             print(f"Train Loss: {train_loss:.4f}, Train Accuracy: {train_accuracy:.4f},  Train F1:{train_f1:.4f}\n")
             print(f"Validation Loss: {val_loss:.4f}, Validation Accuracy: {val_accuracy:.4f}, Validation F1:{val_f1:.4f}")
             self.val_losses.append(val_loss)
             self.val_accuracies.append(val_accuracy)
             self.val_F1.append(val_f1)
             if self.scheduler is not None:
                # Pass the validation loss to the scheduler's step method
                self.scheduler.step(val_loss)

             self.early_stop(val_loss)

             if self.early_stop.early_stop:
                self.early_stop_epoch = epoch
                print("Early stopping")
                break
             self.model_checkpoint(val_loss,self.model,epoch+1)
             self.best_epoch = self.model_checkpoint.best_epoch

        print(f"Finished train at epoch {epoch+1}")

    def train_epoch(self,dataloader: DataLoader) -> tuple[float,float, float]:
      """
      Addestra il modello per un epoca

      Args:
          dataloader (DataLoader): il dataloader per il training set

      Returns:
          tuple[float,float, float]: The loss and accuracy for the epoch.
      """
      self.model.train()
      epoch_accuracy=0.0
      epoch_loss=0.0
      correct=0
      total=0
      all_preds = []
      all_targets = []
      for i, batch in enumerate(dataloader,0):
        images,targets = batch
        self.optimizer.zero_grad()
        images=images.to(device)
        targets=targets.to(device)
        logist=self.model(images)
        loss=self.criterion(logist,targets)
        loss.backward()
        self.optimizer.step()
        epoch_loss +=  loss.item()
        preds = torch.argmax(logist, dim=1)
        correct += (preds == targets).sum().item()
        total += targets.size(0)
        all_preds.extend(preds.cpu().numpy())
        all_targets.extend(targets.cpu().numpy())

        if (i+1) % self.log_every_n_iterations == 0:
            print(f"Step [{i+1}/{len(dataloader)}], Loss: {loss.item():.4f}, Accuracy: {correct/total:.4f}, LR:  {self.optimizer.param_groups[0]['lr']:.6f}")

      self.lr_values.append(self.optimizer.param_groups[0]['lr'])
      epoch_f1_score = f1_score(
        all_targets,
        all_preds,
        average="macro"
      )

      epoch_accuracy = correct / total
      epoch_loss=epoch_loss/len(train_loader)
      return epoch_loss,epoch_accuracy,epoch_f1_score

    def val_epoch(self,dataloader: DataLoader) -> tuple[float,float, float]:
      """
      Valida il modello per l'epoca di train appena conclusa.

      Args:
          dataloader (DataLoader): dataloader del validation set
      Returns:
          tuple[float,float]: Loss, accuracy validation e f1 score per l'epoca.
      """
      self.model.eval()
      epoch_accuracy=0.0
      epoch_loss_total=0.0
      correct=0
      total=0
      all_preds = []
      all_targets = []
      with torch.no_grad():
          for i,batch in enumerate(dataloader,0):
              images,targets = batch
              images=images.to(device)
              targets=targets.to(device)
              logist=self.model(images)
              val_loss=self.criterion(logist,targets)
              epoch_loss_total += val_loss.item()
              preds = torch.argmax(logist, dim=1)
              correct += (preds == targets).sum().item()
              total += targets.size(0)
              all_preds.extend(preds.cpu().numpy())
              all_targets.extend(targets.cpu().numpy())
      epoch_f1_score = f1_score(
        all_targets,
        all_preds,
        average="macro"
      )
      epoch_accuracy = correct / total
      epoch_loss_total=epoch_loss_total/len(dataloader)
      return epoch_loss_total,epoch_accuracy,epoch_f1_score

    def load_model(self,path:str):
      """
      carica il modello da un path
      args:
          path: path del modello
      """
      self.model.load_state_dict(torch.load(path))
      self.model.eval()

    def load_best_model(self):
      """
      carica il modello con i pesi migliori
      """
      self.model.load_state_dict(torch.load(self.model_checkpoint.best_model_path))
      self.model.eval()

    def save_model(self,path:str):
      """
      salva il modello in un path
      args:
          path: path dove salvare il modello
      """
      torch.save(self.model.state_dict(),path)

    def save_history(self,path:str):
      """
      salva le metriche dell'addestramento
      args:
          path: path dove salvare le metriche
      """
      history={
          "train_losses":self.train_losses,
          "val_losses":self.val_losses,
          "train_accuracies":self.train_accuracies,
          "val_accuracies":self.val_accuracies,
          "train_f1":self.train_F1,
          "val_f1":self.val_F1,
          "lr_values":self.lr_values,
          "best_epoch":self.best_epoch,
          "early_stop_epoch":self.early_stop_epoch
      }
      with open(path,"wb") as f:
        pickle.dump(history,f)

    def load_history(self,path:str):
      """
      carica le metriche dell'addestramento
      """
      with open(path,"rb") as f:
        history=pickle.load(f)
        self.train_losses=history["train_losses"]
        self.val_losses=history["val_losses"]
        self.train_accuracies=history["train_accuracies"]
        self.val_accuracies=history["val_accuracies"]
        self.train_F1=history["train_f1"]
        self.val_F1=history["val_f1"]
        self.lr_values=history["lr_values"]
        self.best_epoch = history.get("best_epoch", None)
        self.early_stop_epoch = history.get("early_stop_epoch", None)


    def evaluate_model(self,dataloader:DataLoader,output=False):
      """
      Valida il modello

      Args:
          dataloader (DataLoader): dataloader del validation set
      Returns:
          tuple[float,float]: Loss, accuracy validation e f1 score.
      """
      self.model.eval()
      all_preds = []
      all_labels = []

      with torch.no_grad():
          for images, labels in dataloader:
              images = images.to(self.device)
              labels = labels.to(self.device)
              outputs = self.model(images)
              preds = torch.argmax(outputs, dim=1)
              all_preds.extend(preds.cpu().numpy())
              all_labels.extend(labels.cpu().numpy())

      f1 = f1_score(all_labels, all_preds, average="macro")
      acc = accuracy_score(all_labels, all_preds)
      report = classification_report(all_labels, all_preds, target_names=LABEL_TO_TARGET.keys())
      report_dict = classification_report(all_labels, all_preds, target_names=LABEL_TO_TARGET.keys(), output_dict=True)

      if output:
        print(f"Accuracy: {acc:.4f}")

        print("\nClassification Report:")
        print(report)

      #lista immagini errori
      error_images = []
      error_labels = []
      error_preds = []
      for i in range(len(all_labels)):
        if all_labels[i] != all_preds[i]:
          error_images.append(dataloader.dataset.images[i])
          error_labels.append(all_labels[i])
          error_preds.append(all_preds[i])

      errors={
          "images":error_images,
          "labels":error_labels,
          "preds":error_preds
      }


      return f1, acc, report, report_dict, errors

    def plot_confusion_matrix(self, test_loader:DataLoader, classes:list[str]):
      """
      Plotta la matrice di confusione
      Args:
          test_loader (DataLoader): dataloader del validation set
          classes (list[str]): nomi delle classi
      """
      self.model.eval()
      y_true = []
      y_pred = []

      with torch.no_grad():
          for i, data in enumerate(test_loader, 0):
              images, labels = data[0].to(self.device), data[1].to(self.device)

              outputs = self.model(images)

              _, predicted = torch.max(outputs, 1)

              y_true.extend(labels.cpu().numpy())
              y_pred.extend(predicted.cpu().numpy())

      cm = confusion_matrix(y_true, y_pred)
      plt.figure(figsize=(10,7))
      sns.heatmap(cm, annot=True, fmt='d', xticklabels=classes, yticklabels=classes, cmap='Blues')

    def plot_metrics(self):
      """
      Plotta le metriche di train e validation
      """
      fig, axs = plt.subplots(1, 3, figsize=(15, 5))

      epochs = range(1, len(self.train_losses) + 1)

      # --- LOSS ---
      axs[0].plot(epochs, self.train_losses, label='Train Loss')
      axs[0].plot(epochs, self.val_losses, label='Validation Loss')
      axs[0].set_xlabel('Epoch')
      axs[0].set_ylabel('Loss')
      axs[0].legend()

      # --- ACCURACY ---
      axs[1].plot(epochs, self.train_accuracies, label='Train Accuracy')
      axs[1].plot(epochs, self.val_accuracies, label='Validation Accuracy')
      axs[1].set_xlabel('Epoch')
      axs[1].set_ylabel('Accuracy')
      axs[1].set_ylim(0, 1)
      axs[1].legend()

      # --- F1 ---
      axs[2].plot(epochs, self.train_F1, label='Train F1')
      axs[2].plot(epochs, self.val_F1, label='Validation F1')
      axs[2].set_xlabel('Epoch')
      axs[2].set_ylabel('F1')
      axs[2].set_ylim(0, 1)
      axs[2].legend()

      if hasattr(self, "early_stop_epoch"):
        for ax in axs:
            ax.axvline(
                x=self.early_stop_epoch+1,
                linestyle="--",
                color="red",
                label="Early stopping"
            )
            ax.legend()
      if hasattr(self, "best_epoch"):
        for ax in axs:
            ax.axvline(
                x=self.best_epoch+1,
                color="green",
                linestyle=":",
                linewidth=2,
                label="Best epoch"
            )

      plt.tight_layout()
      plt.show()

    def plot_lr(self):
      """
      Plotta il valore di learning rate
      """
      epochs = range(1, len(self.train_losses) + 1)

      plt.figure(figsize=(10, 5))
      plt.plot(epochs, self.lr_values, label='Learning Rate')
      plt.xlabel('Epoch')
      plt.ylabel('Learning Rate')
      plt.legend()
      plt.show()
```

```python
def show_errors(errors:dict,classes:list[str]):
  """
  funzione che mostra le immagini con classificazione errata
  args:
        errors (dictionary): immagini degli errori
        classes (list[str]): lista delle classi
  """
  mis_images=errors["images"]
  mis_labels=errors["labels"]
  mis_preds=errors["preds"]

  n_errors = len(mis_images)
  num_col=5
  num_row = (n_errors + num_col - 1) // num_col

  fig, axes = plt.subplots(num_row, num_col, figsize=(4 * num_col, 4 * num_row))
  for i in range(len(mis_images)):
      ax = axes[i]
      image = Image.open(mis_images[i]).convert("RGB")
      # denormalizziamo
      ax.imshow(image)
      ax.axis('off')
      ax.set_title(f"T:{classes[mis_labels[i]]}/P:{classes[mis_preds[i]]}",fontsize=12)

  # Nascondo gli assi inutilizzati
  for j in range(n_errors, len(axes)):
      axes[j].axis("off")

  plt.tight_layout()
  plt.show()
```

```python
def show_errors_gradcam(
    model:nn.Module,
    errors: dict,
    classes: list[str],
    transform: callable,
    target_layer: nn.Module,
    device: torch.device,
    max_images: int = 10
):

    """
    Funzione che mostra le immagini con classificazione errata con GradCAM
    Args:
        model (nn.Module): modello di classificazione
        errors (dict): dizionario contenente le immagini degli errori
        classes (list[str]): lista delle classi
        transform (callable): trasformazione
        target_layer (nn.Module): layer di target per GradCAM
        device (torch.device): dispositivo su cui eseguire il calcolo
        max_images (int): numero massimo di immagini da mostrare
    """
    model.eval()

    cam = GradCAM(
        model=model,
        target_layers=[target_layer]
    )

    mis_images = errors["images"][:max_images]
    mis_labels = errors["labels"][:max_images]
    mis_preds  = errors["preds"][:max_images]

    num_col = 5
    num_row = int(np.ceil(len(mis_images) / num_col))

    fig, axes = plt.subplots(num_row, num_col, figsize=(4*num_col, 4*num_row))
    axes = axes.flatten()

    # resize immagine iniziale.
    resize_for_cam_display = T.Resize((224, 224))

    for i, img_path in enumerate(mis_images):
        image_original_pil = Image.open(img_path).convert("RGB")
        input_tensor = transform(image_original_pil).unsqueeze(0).to(device)

        # Classe predetta (errore!)
        targets = None  # default = classe predetta

        grayscale_cam = cam(
            input_tensor=input_tensor,
            targets=targets
        )[0]


        image_for_cam_display_pil = resize_for_cam_display(image_original_pil)
        rgb_img_for_display = np.array(image_for_cam_display_pil).astype(np.float32) / 255.0 #normalizzazione

        cam_image = show_cam_on_image(rgb_img_for_display, grayscale_cam, use_rgb=True)

        ax = axes[i]
        ax.imshow(cam_image)
        ax.axis("off")
        ax.set_title(
            f"T:{classes[mis_labels[i]]}\nP:{classes[mis_preds[i]]}",
            fontsize=12
        )

    for j in range(i+1, len(axes)):
        axes[j].axis("off")

    plt.tight_layout()
    plt.show()
```

##Trasformazioni

Definiamo la traformazione per le immagini del train set.
Le immagini verranno ridimensionate in 224x224 per facilitare le reti neurali, verrà predisposto un flip orizzontale e una rotazione casuale di 10 gradi, inoltre verranno effettuate una sfocatura gaussiana, per intercettare anche foto non perfettamente a fuoco e un ColorJitter che modificherà lievemente contrasto, saturazione, lumonisità e tonalità della foto per variarne leggermente i colori.
Infine l'immagine verrà convertita in tensore e normalizzata.
Dato che utilizzeremo **modelli pretrained su ImageNet** dovremo usare la medisima normalizzazione prevista per Image.net

```python
train_transform = T.Compose([
    T.RandomResizedCrop(224, scale=(0.8, 1.0)),
    T.RandomHorizontalFlip(),
    T.RandomRotation(10),
    T.RandomApply([
        T.GaussianBlur(kernel_size=3, sigma=(0.1, 1.0))
    ], p=0.2),
    T.ColorJitter(brightness=0.2, contrast=0.2, saturation=0.2,hue=0.03),
    T.ToTensor(),
    T.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225]),
])
```

Definiamo la trasformazione per le immagini del test e validation set.

```python
val_transform = T.Compose([
    T.Resize((224,224)),
    T.ToTensor(),
    T.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225]),
])
```
